```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import numpy as np
from typing import List, Union

load_dotenv()

def genrate_embeddings(texts: Union[str, List[str]], model: str = "text-embedding-3-small") -> List[np.ndarray]:
    client = OpenAI(
        api_key=os.environ["EURI_API_KEY"],
        base_url="https://api.euron.one/api/v1/euri",
    )
    resp = client.embeddings.create(
        model=model,
        input=texts,
    )
    return [np.array(item.embedding) for item in resp.data]
```

```python
index_name = "doc-index"

if not pc.has_index(index_name):
    pc.create_index(
        name=index_name,
        dimension=1536,  # specify the embedding dimension for the model
        metric="cosine",  # specify the similarity metric for the index
        spec = ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
```

```python
texts = [doc["text"] for doc in documents]
```

```python
texts
```

```python
doc_embeddings = genrate_embeddings(texts)
```

```python
doc_embeddings
```

```python
vectors_to_upsert = []
for doc, emb in zip(documents, doc_embeddings):
    vectors_to_upsert.append({"id": doc["id"], "values": emb.tolist(), "metadata": dict(doc["metadata"], **{"texts": doc["text"]}), })
```

```python
vectors_to_upsert[0]
```

```python
pc.index(index_name).upsert(vectors=vectors_to_upsert)
```

```python
# Write the question we want to search for.
query_text = "Can We use Pinecone with OpenAI embeddings?"

# Convert the question into a vector using the same embedding model as the documents.
query_vector = genrate_embeddings(query_text)[0].tolist()

# Ask Pinecone for the 3 closest documents and include their metadata.
search_response = pc.index(index_name).query(
    vector=query_vector,
    top_k=3,
    include_metadata=True,
)

# Display the matching documents.
search_response["matches"]
```
